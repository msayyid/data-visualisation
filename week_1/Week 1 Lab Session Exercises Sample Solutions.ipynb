```python
import matplotlib.pyplot as plt
import numpy as np

# Create data
x = np.linspace(0, 2*np.pi, 100)
y1 = 4 + 2 * np.sin(2 * x)
y2 = 4 + 2 * np.cos(2 * x)

# Create figure and axis
fig, ax = plt.subplots(figsize=(12, 8))

# Set custom colors for a pastel-like appearance
pastel_colors = ['#FFA07A', '#98FB98', '#87CEFA', '#DDA0DD', '#F0E68C']
plt.rcParams['axes.prop_cycle'] = plt.cycler(color=pastel_colors)

# Plot data
line1, = ax.plot(x, y1, label='sin', linewidth=2)
line2, = ax.plot(x, y2, label='cos', linewidth=2)

# Set title and labels
ax.set_title("Anatomy of a Matplotlib Figure", fontsize=16, fontweight='bold')
ax.set_xlabel("x (radians)", fontsize=12)
ax.set_ylabel("y", fontsize=12)

# Exercise 1: Customize Tick Labels
ax.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
ax.set_xticklabels(['0', r'$\pi/2$', r'$\pi$', r'$3\pi/2$', r'$2\pi$'])

# Exercise 2: Add Grid
ax.grid(which='major', linestyle='--', linewidth=1.2, color='gray', alpha=0.7)
ax.grid(which='minor', linestyle=':', linewidth=0.8, color='gray', alpha=0.4)
ax.set_axisbelow(True)  # Place gridlines behind plot elements
ax.minorticks_on()

# Exercise 3: Customize Spines
ax.spines['left'].set_position(('data', 0))
ax.spines['bottom'].set_position(('data', 0))
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Exercise 4: Add Text Annotation
max_sin_x = np.pi/4
max_sin_y = 4 + 2 * np.sin(2 * max_sin_x)
ax.annotate('Max of sin wave', xy=(max_sin_x, max_sin_y), xytext=(max_sin_x+0.5, max_sin_y+0.5),
            arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=8),
            fontsize=10, ha='left', va='bottom')

# Exercise 5: Add a Secondary Y-axis
ax2 = ax.twinx()
ax2.set_ylabel('y (in π units)', fontsize=12)
ax2.set_ylim(ax.get_ylim())
ax2.set_yticks(np.linspace(ax.get_ylim()[0], ax.get_ylim()[1], 5))
ax2.set_yticklabels([f'{y/np.pi:.2f}π' for y in ax2.get_yticks()])

# Exercise 7: Add Inset Axes
axins = ax.inset_axes([0.55, 0.05, 0.3, 0.3])
axins.plot(x, y1, x, y2)
axins.set_xlim(3*np.pi/4, 5*np.pi/4)
axins.set_ylim(3, 5)
axins.set_xticklabels([])
axins.set_yticklabels([])
axins.set_facecolor('#F0F0F0')  # Light gray background for inset
ax.indicate_inset_zoom(axins, edgecolor="black")

# Add legend
ax.legend(loc='upper right', fontsize=10, facecolor='white', edgecolor='none', framealpha=0.7)

# Customize background
ax.set_facecolor('#F8F8F8')  # Very light gray background
fig.patch.set_facecolor('white')

# Adjust layout
plt.tight_layout()

# Show plot
plt.show()
fig.savefig('Anatomy of a Matplotlib Figure.png')
# Detailed comments for each exercise:

# Exercise 1: Customize Tick Labels
# We use set_xticks to specify the positions of the ticks (in data coordinates)
# set_xticklabels is used to provide custom labels for these ticks
# The r before the string allows us to use LaTeX formatting for pi symbols

# Exercise 2: Add Grid
# We add two types of grids: major and minor
# The major grid uses dashed lines (--) and is darker
# The minor grid uses dotted lines (:) and is lighter
# set_axisbelow(True) ensures the grid is drawn behind other plot elements
# minorticks_on() is called to enable minor ticks

# Exercise 3: Customize Spines
# We move the left and bottom spines to x=0 and y=0 respectively
# The top and right spines are made invisible
# This creates a plot that looks more like a traditional x-y coordinate system

# Exercise 4: Add Text Annotation
# We calculate the maximum point of the sine wave
# annotate() is used to add text with an arrow pointing to this maximum
# xytext specifies where the text should be placed relative to the point

# Exercise 5: Add a Secondary Y-axis
# twinx() creates a second y-axis that shares the same x-axis
# We set custom tick locations and labels to show the values in terms of π

# Exercise 6: Customize Figure Style
# We set custom colors and styles
# We use pastel colors for the lines and customize the background colors

# Exercise 7: Add Inset Axes
# inset_axes() creates a smaller axes inside the main plot
# We plot the same data but set the limits to zoom in on the intersection
# indicate_inset_zoom() adds connecting lines between the inset and the main plot
# We set a light gray background for the inset to make it stand out

# Note: This solution combines all exercises into one comprehensive figure,
# addressing the bonus challenge of creating a single figure that showcases
# multiple aspects of Matplotlib's anatomy.
```
